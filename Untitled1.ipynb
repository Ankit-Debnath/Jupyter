````python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
````

````python
df = pd.read_csv("Financial_Ecosystem_Dataset_T1_5k.csv")

df.columns = df.columns.str.strip()

print("Dataset Shape:", df.shape)

df.head()
````

````python
print("Dataset Information:")
df.info()
````

````python
print("Total Missing Values:", df.isnull().sum().sum())

print("\nMissing Values by Column:")
print(
    df.isnull().sum()[
        df.isnull().sum() > 0
    ]
)
````

````python
print("Duplicate Rows:", df.duplicated().sum())
````

````python
print("Target Distribution:")
print(df["Target_Credit_Default"].value_counts())

print("\nTarget Percentage:")
print(
    df["Target_Credit_Default"]
    .value_counts(normalize=True)
    .mul(100)
)
````

````python
df["Account_Open_Date"] = pd.to_datetime(
    df["Account_Open_Date"],
    errors="coerce"
)

df["Txn_Timestamp"] = pd.to_datetime(
    df["Txn_Timestamp"],
    errors="coerce"
)
````

````python
df["Account_Open_Year"] = df["Account_Open_Date"].dt.year
df["Account_Open_Month"] = df["Account_Open_Date"].dt.month

df["Txn_Hour"] = df["Txn_Timestamp"].dt.hour
df["Txn_Day"] = df["Txn_Timestamp"].dt.day
df["Txn_Month"] = df["Txn_Timestamp"].dt.month
df["Txn_DayOfWeek"] = df["Txn_Timestamp"].dt.dayofweek

df["Account_Age_Days"] = (
    df["Txn_Timestamp"] - df["Account_Open_Date"]
).dt.days
````

````python
df = df.drop(
    columns=[
        "Account_Open_Date",
        "Txn_Timestamp"
    ]
)

print("Shape after date processing:", df.shape)
````

````python
df = df.drop(
    columns=[
        "Delinquency_Status",
        "Risk_Score_Internal"
    ],
    errors="ignore"
)

print("Shape after removing leakage-prone features:", df.shape)
````

````python
target = "Target_Credit_Default"

y = df[target].astype(int)

X = df.drop(
    columns=[
        "Target_Credit_Default",
        "Target_Is_Fraud_AML"
    ]
)

print("X Shape:", X.shape)
print("y Shape:", y.shape)
````

````python
id_columns = [
    "Cust_ID",
    "Account_ID",
    "Txn_ID",
    "Counterparty_ID"
]

X = X.drop(
    columns=id_columns,
    errors="ignore"
)

print("Final X Shape:", X.shape)
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape :", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
````

````python
categorical_features = X_train.select_dtypes(
    include=["object", "bool"]
).columns.tolist()

numerical_features = X_train.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

print("Number of categorical features:", len(categorical_features))
print(categorical_features)

print("\nNumber of numerical features:", len(numerical_features))
print(numerical_features)
````

````python
numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)
print("Numerical preprocessing pipeline created.")
````

````python
categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            )
        )
    ]
)
print("Categorical preprocessing pipeline created.")
````

````python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numeric_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)
````

````python
logistic_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            LogisticRegression(
                class_weight="balanced",
                max_iter=2000,
                random_state=42
            )
        )
    ]
)
````

````python
logistic_model.fit(
    X_train,
    y_train
)

print("Logistic Regression training completed.")
````

````python
y_pred_lr = logistic_model.predict(X_test)

y_prob_lr = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions completed.")
print("Predicted classes:", y_pred_lr[:10])
print("Predicted probabilities:", y_prob_lr[:10])
````

````python
random_forest_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=10,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
````

````python
random_forest_model.fit(
    X_train,
    y_train
)

print("Random Forest training completed.")
````

````python
y_pred_rf = random_forest_model.predict(X_test)

y_prob_rf = random_forest_model.predict_proba(X_test)[:, 1]

print("Predictions completed.")
print("Predicted classes:", y_pred_rf[:10])
print("Predicted probabilities:", y_prob_rf[:10])
````

````python
print("Number of predicted defaults:", y_pred_rf.sum())
print("Number of actual defaults:", y_test.sum())

print("\nHighest predicted default probabilities:")
print(np.sort(y_prob_rf)[-10:])
````

````python
decision_tree_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            DecisionTreeClassifier(
                max_depth=10,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)
````

````python
decision_tree_model.fit(
    X_train,
    y_train
)

print("Decision Tree training completed.")
````

````python
y_pred_dt = decision_tree_model.predict(X_test)

y_prob_dt = decision_tree_model.predict_proba(X_test)[:, 1]

print("Predictions completed.")
print("Predicted classes:", y_pred_dt[:10])
print("Predicted probabilities:", y_prob_dt[:10])
````

````python
def evaluate_model(name, y_true, y_pred, y_prob):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_true, y_prob)

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")
    print(f"ROC-AUC   : {roc_auc:.4f}")

    print("\nClassification Report:")
    print(classification_report(
        y_true,
        y_pred,
        target_names=["No Default", "Default"],
        zero_division=0
    ))
````

````python
evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_lr,
    y_prob_lr
)

evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf,
    y_prob_rf
)

evaluate_model(
    "Decision Tree",
    y_test,
    y_pred_dt,
    y_prob_dt
)
````

````python
models = [
    ("Logistic Regression", y_pred_lr, y_prob_lr),
    ("Random Forest", y_pred_rf, y_prob_rf),
    ("Decision Tree", y_pred_dt, y_prob_dt)
]

results = []

for name, y_pred, y_prob in models:
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

results_df
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

models_cm = [
    ("Logistic Regression", y_pred_lr),
    ("Random Forest", y_pred_rf),
    ("Decision Tree", y_pred_dt)
]

for ax, (name, y_pred) in zip(axes, models_cm):
    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Default", "Default"],
        yticklabels=["No Default", "Default"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
````

````python
metrics = ["Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"]

results_plot = results_df.set_index("Model")[metrics]

ax = results_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=0)
plt.legend(title="Metrics")
plt.tight_layout()
plt.show()
````

````python
best_f1 = results_df["F1 Score"].max()

best_models = results_df.loc[
    results_df["F1 Score"] == best_f1,
    "Model"
].tolist()

print("Best F1 Score:", best_f1)
print("Models achieving the best F1 Score:")

for model in best_models:
    print("-", model)
````

````python
from sklearn.metrics import average_precision_score

pr_results = []

for name, y_pred, y_prob in models:
    pr_auc = average_precision_score(y_test, y_prob)

    pr_results.append({
        "Model": name,
        "PR-AUC": pr_auc
    })

pr_auc_df = pd.DataFrame(pr_results)

print(pr_auc_df)
````

````python
# Get the fitted preprocessing transformer
rf_preprocessor = random_forest_model.named_steps["preprocessor"]

# Get the fitted Random Forest classifier
rf_classifier = random_forest_model.named_steps["classifier"]


# -----------------------------------------
# 1. Numerical feature names
# -----------------------------------------
num_features = numerical_features


# -----------------------------------------
# 2. Get categorical pipeline
# -----------------------------------------
cat_pipeline = rf_preprocessor.named_transformers_["categorical"]


# -----------------------------------------
# 3. Get fitted OneHotEncoder
# -----------------------------------------
onehot_encoder = cat_pipeline.named_steps["onehot"]


# -----------------------------------------
# 4. Get encoded categorical feature names
# -----------------------------------------
cat_features = onehot_encoder.get_feature_names_out(
    categorical_features
)


# -----------------------------------------
# 5. Combine all feature names
# -----------------------------------------
feature_names = np.concatenate([
    num_features,
    cat_features
])


# -----------------------------------------
# 6. Get Random Forest feature importance
# -----------------------------------------
feature_importances = rf_classifier.feature_importances_


# -----------------------------------------
# 7. Create feature importance DataFrame
# -----------------------------------------
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": feature_importances
})


# Sort from most important to least important
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)


# Display top 20 features
feature_importance_df.head(20)
````

````python
# Select top 15 features
top_features = feature_importance_df.head(15)

# Create horizontal bar chart
plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Top 15 Random Forest Feature Importances")

plt.tight_layout()
plt.show()
````

# Detailed Explanation of `Untitled1.ipynb`

## 1. Project Overview

This notebook implements an end-to-end **Credit Default Prediction**
machine-learning workflow using a financial ecosystem transaction
dataset.

The overall plan is:

1.  Load and inspect the financial dataset.
2.  Understand data types, missing values, duplicates, and class
    distribution.
3.  Convert date/time columns into useful numerical features.
4.  Remove variables that may cause target leakage.
5.  Remove identifiers and unrelated target columns.
6.  Split the transaction data into training and testing sets while
    preserving the rare default class.
7.  Build separate preprocessing pipelines for numerical and categorical
    data.
8.  Train three classification models:
    -   Logistic Regression
    -   Random Forest
    -   Decision Tree
9.  Evaluate the models using several classification metrics.
10. Investigate the unusually perfect scores.
11. Measure PR-AUC because the target is highly imbalanced.
12. Analyze Random Forest feature importance.
13. Document the dataset limitations and avoid interpreting the perfect
    scores as real-world 100% accuracy.

------------------------------------------------------------------------

------------------------------------------------------------------------

# 9. How We Deal With Class Imbalance

The notebook uses **two important strategies**.

## 9.1 Stratified train-test split

The split is:

``` python
train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

`stratify=y` attempts to preserve the class proportions in both sets.

The resulting distribution is:

### Training

``` text
No Default = 3980
Default    = 20
```

### Testing

``` text
No Default = 995
Default    = 5
```

So the rare positive class is represented in both sets.

------------------------------------------------------------------------

## 9.2 Class weighting

All three classifiers use:

``` python
class_weight="balanced"
```

This is a key part of the imbalance strategy.

Instead of generating synthetic examples, the models give greater
importance to the minority class during training.

Conceptually:

``` text
Majority class
    ↓
Lower relative weight

Minority class
    ↓
Higher relative weight
```

This encourages the model not to simply ignore the rare default cases.

### Why we did not use SMOTE

The notebook does not use SMOTE.

This is appropriate for the current dataset because the main problem is
deeper than simply having too few positive rows.

All 25 default transactions belong to **one customer**.

Synthetic oversampling would create additional synthetic examples based
on essentially one customer's default pattern. That would not create
genuine independent defaulting customers.

Therefore, class weighting is a more defensible choice for this
notebook.

# 11. Leakage-Prone Features

Two columns are deliberately removed:

``` text

Risk_Score_Internal
```

The reason is that these variables may contain information that is too
directly connected to the outcome.

For example, delinquency status can essentially describe whether a
customer is already seriously delinquent.

Using such a variable to predict default can make the model appear much
better than it would be in a genuine prediction scenario.

Therefore, they were excluded.

# 21. Model Evaluation

The notebook evaluates:

### Accuracy

Percentage of all predictions that are correct.

### Precision

Among predicted defaults, how many were actually defaults?

### Recall

Among actual defaults, how many were detected?

Recall is especially important in default prediction because missing a
genuine default can be costly.

### F1 Score

Harmonic mean of precision and recall.

### ROC-AUC

Measures how well the model ranks positive cases above negative cases
across classification thresholds.

### PR-AUC

Especially useful for imbalanced classification because it focuses on
the precision-recall relationship.

# 23. Why the 100% Scores Need Caution

The most important analytical discovery in the notebook is that all 25
default transactions belong to exactly one customer:

``` text
CUST-00000284
```

That customer has:

``` text
25 transactions
25 defaults
```

The train-test split is performed at the **transaction level**, not at
the customer level.

Therefore, transactions from the same customer can occur in both
training and testing.

The default customer is present in both sets.

This means the test set is not a truly independent set of unseen
customers.

The models can learn patterns associated with this particular customer's
profile and then encounter the same customer again in the test set.

Therefore:

> The 1.00 metrics describe performance on this transaction-level test
> split, but they should not be interpreted as evidence of 100%
> generalization to new customers.

------------------------------------------------------------------------

# 24. Why a Customer-Level Split Is Difficult Here

For a new-customer prediction problem, the preferred methodology would
be:

``` text
Customer A → Train only
Customer B → Train only
Customer C → Test only
Customer D → Test only
```

No customer should occur in both sets.

However, this dataset contains only **one customer with any default
cases**.

If that customer is placed in training:

``` text
Training → has positive cases
Testing  → has zero positive cases
```

If that customer is placed in testing:

``` text
Training → has zero positive cases
Testing  → has all positive cases
```

Therefore, a meaningful customer-level classification experiment cannot
be performed with this particular dataset.

The appropriate solution for a stronger project would be a dataset
containing defaults from many independent customers.
